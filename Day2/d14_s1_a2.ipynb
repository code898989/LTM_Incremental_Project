```python

import numpy as np

from sklearn.datasets import load_breast_cancer
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score


# 1. Load Data
def load_data():
    d = load_breast_cancer()

    X = d.data
    y = d.target
    feature_names = list(d.feature_names)

    return X, y, feature_names


# 2. Fit Decision Tree
def fit_tree(X, y, max_depth=None):

    model = DecisionTreeClassifier(
        max_depth=max_depth,
        random_state=0
    )

    model.fit(X, y)

    return model


# 3. Feature Importance Ranking
def importance_ranking(model, feature_names):

    importances = model.feature_importances_

    ranking = list(zip(feature_names, importances))

    ranking.sort(key=lambda x: x[1], reverse=True)

    return [
        (name, round(float(importance), 4))
        for name, importance in ranking
    ]


# 4. Train Test Accuracy
def train_test_accuracy(X, y, max_depth):

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.30,
        random_state=0,
        stratify=y
    )

    model = DecisionTreeClassifier(
        max_depth=max_depth,
        random_state=0
    )

    model.fit(X_train, y_train)

    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    train_acc = accuracy_score(y_train, y_train_pred)
    test_acc = accuracy_score(y_test, y_test_pred)

    return round(train_acc, 3), round(test_acc, 3)


# 5. Overfitting Gap
def overfit_gap(X, y, max_depth):

    train_acc, test_acc = train_test_accuracy(
        X, y, max_depth
    )

    return round(train_acc - test_acc, 3)


# Verification
X, y, names = load_data()

print("Data Shape:", X.shape)
print("Target Shape:", y.shape)
print("Number of Features:", len(names))

full = fit_tree(X, y, None)

print("\nTop Feature:")
print(importance_ranking(full, names)[0])

print("\nFeature Importance Ranking:")
print(importance_ranking(full, names))

print("\nFull Tree Accuracy:")
print(train_test_accuracy(X, y, None))

print("\nDepth 1 Accuracy:")
print(train_test_accuracy(X, y, 1))

print("\nDepth 3 Accuracy:")
print(train_test_accuracy(X, y, 3))

print("\nOverfitting Gap:")
print("Full Tree:", overfit_gap(X, y, None))
print("Depth 1:", overfit_gap(X, y, 1))
print("Depth 3:", overfit_gap(X, y, 3))
```
