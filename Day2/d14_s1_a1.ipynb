```python

import numpy as np


# 1. Gini Impurity
def gini(y):
    y = np.array(y)

    if len(y) == 0:
        return 0.0

    _, counts = np.unique(y, return_counts=True)
    probabilities = counts / len(y)

    return 1.0 - np.sum(probabilities ** 2)


# 2. Entropy
def entropy(y):
    y = np.array(y)

    if len(y) == 0:
        return 0.0

    _, counts = np.unique(y, return_counts=True)
    probabilities = counts / len(y)

    return -np.sum(probabilities * np.log2(probabilities))


# 3. Information Gain
def information_gain(y_parent, y_left, y_right, criterion="gini"):

    if criterion == "gini":
        impurity = gini
    elif criterion == "entropy":
        impurity = entropy
    else:
        raise ValueError("Invalid criterion")

    n = len(y_parent)

    if n == 0:
        return 0.0

    parent_impurity = impurity(y_parent)

    weighted_child_impurity = (
        (len(y_left) / n) * impurity(y_left)
        + (len(y_right) / n) * impurity(y_right)
    )

    return parent_impurity - weighted_child_impurity


# 4. Best Threshold
def best_threshold(x, y, criterion="gini"):

    x = np.array(x)
    y = np.array(y)

    if len(x) != len(y):
        raise ValueError("x and y must have the same length")

    if len(x) < 2 or len(np.unique(x)) < 2:
        return None, 0.0

    unique_values = np.unique(x)

    thresholds = (
        unique_values[:-1] + unique_values[1:]
    ) / 2

    best_t = None
    best_gain = -float("inf")

    for t in thresholds:

        left_mask = x <= t
        right_mask = x > t

        y_left = y[left_mask]
        y_right = y[right_mask]

        gain = information_gain(
            y, y_left, y_right, criterion
        )

        if gain > best_gain:
            best_gain = gain
            best_t = t

    return best_t, round(best_gain, 4)


# Verification
y_parent = np.array([0, 0, 0, 1, 1, 1])
x = np.array([0.1, 0.2, 0.9, 1.0, 1.1])
y = np.array([0, 0, 1, 1, 1])

print(gini([0, 0, 1, 1]))
print(gini([1, 1, 1, 1]))

print(entropy([0, 0, 1, 1]))
print(entropy([1, 1, 1, 1]))

print(information_gain(
    [0, 0, 1, 1],
    [0, 0],
    [1, 1],
    criterion="gini"
))

print(information_gain(
    [0, 0, 1, 1],
    [0, 0],
    [1, 1],
    criterion="entropy"
))

print(best_threshold(
    [1, 2, 3, 4],
    [0, 0, 1, 1],
    criterion="gini"
))

print(best_threshold(
    [1, 2, 3, 4],
    [0, 0, 1, 1],
    criterion="entropy"
))

print(best_threshold(x, y, criterion="gini"))
print(best_threshold(x, y, criterion="entropy"))
```
