```
import numpy as np
from scipy.stats import rankdata

def confusion_at(y_true, y_scores, threshold):
    pred = (np.asarray(y_scores) >= threshold).astype(int)
    y_true = np.asarray(y_true)
    tp = int(np.sum((y_true==1)&(pred==1)))
    fp = int(np.sum((y_true==0)&(pred==1)))
    tn = int(np.sum((y_true==0)&(pred==0)))
    fn = int(np.sum((y_true==1)&(pred==0)))
    return {'tp':tp,'fp':fp,'tn':tn,'fn':fn}

def precision_recall_f1(tp, fp, tn, fn):
    p = tp/(tp+fp) if (tp+fp) else 0.0
    r = tp/(tp+fn) if (tp+fn) else 0.0
    f1 = 2*p*r/(p+r) if (p+r) else 0.0
    return {'precision':float(p),'recall':float(r),'f1':float(f1)}

def roc_auc(y_true, y_scores):
    y_true=np.asarray(y_true)
    ranks=rankdata(y_scores)
    npos=np.sum(y_true==1)
    nneg=np.sum(y_true==0)
    return float((ranks[y_true==1].sum()-npos*(npos+1)/2)/(npos*nneg))

def best_threshold(y_true, y_scores, thresholds):
    best_t=thresholds[0]; best_f=-1
    for t in thresholds:
        c=confusion_at(y_true,y_scores,t)
        f=precision_recall_f1(**c)['f1']
        if f>best_f:
            best_f=f; best_t=t
    return best_t
```
